```python
import numpy as np
import cv2
from tensorflow.keras.models import load_model

```

```python
# Load the pre-trained models
model_isl = load_model('isl_model.keras')
model_asl = load_model('asl_model.keras')
model_gsl = load_model('gsl_model.keras')
```

```python
# Define image dimensions
img_height, img_width = 64, 64  # Use the same dimensions as used during training
```

```python
# Real-time detection through webcam
def real_time_detection(model, language_name):
    # Start capturing from the webcam
    cap = cv2.VideoCapture(0)
    cap.set(3, 640)  # Set width
    cap.set(4, 480)  # Set height

    while True:
        ret, frame = cap.read()
        if not ret:
            print("Failed to grab frame")
            break

        # Preprocess the frame for your model
        input_frame = cv2.resize(frame, (img_height, img_width))  # Adjust input size
        input_frame = np.expand_dims(input_frame, axis=0)  # Add batch dimension
        input_frame = input_frame / 255.0  # Normalize if required

        # Make predictions
        predictions = model.predict(input_frame)
        predicted_class = np.argmax(predictions, axis=1)

        # Display the predicted class on the frame
        cv2.putText(frame, f'Predicted: {predicted_class[0]}', (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 0, 0), 2)
        cv2.putText(frame, f'Language: {language_name}', (10, 60), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)

        # Show the frame
        cv2.imshow('Webcam', frame)

        # Break the loop on 'q' key press
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    # Release the capture and close windows
    cap.release()
    cv2.destroyAllWindows()

```

```python
# User input for sign language selection
print("Select the sign language model to use:")
print("1 -> Indian Sign Language (ISL)")
print("2 -> German Sign Language (GSL)")
print("3 -> American Sign Language (ASL)")

choice = input("Enter your choice (1/2/3): ")

if choice == '1':
    real_time_detection(model_isl, "Indian Sign Language")
elif choice == '2':
    real_time_detection(model_gsl, "German Sign Language")
elif choice == '3':
    real_time_detection(model_asl, "American Sign Language")
else:
    print("Invalid choice. Please select 1, 2, or 3.")
```
